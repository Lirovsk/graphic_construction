```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress
```

```python
data = pd.read_csv('data.csv')

print(data.head())
```

## Ploting the data

The following cell will plot the graphic corresponding to the actual data. This graphic does not contain the linear regression of the data.

```python
data.plot(x='temp', y='V_value')

plt.show()
```

## Linear regression

The next cell will calculate the linear regression of the empirical data saved on `data`.

```python
result = linregress(data['temp'], data['V_value'])

slope = result.slope
intercept = result.intercept
determination_coefficient = result.rvalue ** 2

print(f"Equation: Tension = {slope:.4}*temperature + {intercept:.4}")
print(f"Determination coefficient: {determination_coefficient:.4}")
```

## Comparison graphic

The cell below will plot a graphic containing both the empirical data and the treated data.

```python
plt.plot(data['temp'], data['V_value'], 'o', label='Untreated data', markersize=1)
plt.plot(data['temp'], slope * data['temp'] + intercept, '-', label='Linear regression of the data', linewidth=1)
plt.xlabel('Temperature (°C)')
plt.ylabel('Tension (V)')
plt.title('Linear Regression of Tension vs Temperature')
plt.legend()
plt.show()
```
